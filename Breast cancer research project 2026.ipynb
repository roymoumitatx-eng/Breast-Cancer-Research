```python
# Load a standard public cancer database repository (e.g., Breast Cancer Wisconsin)
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target) # 0 = Malignant, 1 = Benign
```

```python
print(f"Dataset shape: {X.shape}")
```

```python
# Data Cleaning & Preprocessing
# Check for missing values & clean
X = X.dropna()
```

```python
# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Split data into training and test datasets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(name='precision'), tf.keras.metrics.Recall(name='recall')]
)
```

```python
# 4. Evaluate Test Set Performance
test_loss, test_acc, test_precision, test_recall = model.evaluate(X_test, y_test)
print(f"\n✅ Model Training Complete!")
print(f"Test Accuracy: {test_acc * 100:.2f}% | Test Precision: {test_precision * 100:.2f}%")
```
